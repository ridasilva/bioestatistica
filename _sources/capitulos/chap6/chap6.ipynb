# 6 - Variáveis Aleatórias Contínuas

## Introdução

Em probabilidade, uma função $X$ que associa a cada evento do espaço amostral um número real $X(\omega)\in\mathbb{R}$ é denominada uma **variável aleatória** (VA).

Uma variável aleatória pode ser classificada como **discreta** ou **contínua**, dependendo do domínio dos valores de $X$.

Exemplo: o número de alunos em uma sala é uma variável aleatória (discreta), denotada por $X$ (maiúsculo). Uma observação dessa variável é denotada pela respectiva letra minúscula, *e.g.*, $x = 50$ alunos.

Em geral, denotamos a probabilidade de uma V.A. $X$ assumir determinado valor $x$ como $P[X]$ ou $P[X = x]$.

## Distribuições de probabilidade

Existem diversos *modelos probabilísticos* que procuram descrever vários tipos de variáveis aleatórias: são as **distribuições de probabilidade de variáveis aleatórias** (discretas ou contínuas).

A distribuição de probabilidades de uma VA $X$ é, portanto, uma descrição das probabilidades associadas com os possíveis valores de $X$. Os valores que $X$ assume determinam o **suporte** (S) da VA.

- **Variáveis discretas** → suporte em um conjunto de valores enumeráveis (finitos ou infinitos)  
- **Variáveis contínuas** → suporte em um conjunto não enumerável de valores

---

Denomina-se de **distribuição de probabilidade** de alguma variável aleatória, a **regra** geral que define a

- **função de probabilidade** (fp) (V.A.s discretas), ou a  
- **função densidade de probabilidade** (fdp) (V.A.s contínuas)

para a variável de interesse.

Existem muitas distribuições de probabilidade, mas algumas merecem destaque por sua importância prática. Estas distribuições também são chamadas de **modelos probabilísticos**.

---

## Variáveis aleatórias contínuas

Uma V.A. é classificada como contínua se assume valores em qualquer intervalo dos números reais, ou seja, um conjunto de valores não enumerável. Dessa forma, não é possível atribuir probabilidades para um ponto específico, apenas para intervalos da reta.

Exemplos:

- Peso de animais  
- Tempo de falha de um equipamento eletrônico  
- Altura da maré em uma hora específica  
- Salinidade da água do mar  
- Retorno financeiro de um investimento

## Área sob uma curva [1]

Vamos aproximar a i-ésima faixa $S_i$ por um retângulo com largura $\Delta x$ e altura $f(x_i)$, que é o valor de $f$ na extremidade direita. Então, a área do i-ésimo retângulo é $f(x_i)\Delta x$. O que consideramos intuitivamente como a área de $S$ é aproximado pela soma das áreas desses retângulos:

![stewart illustration](figs/stewart_5_11_a)

$$
A = f(x_1)\Delta x+f(x_2)\Delta x+\ldots+f(x_n)\Delta x
$$

---

## Área sob uma curva — Definição de Integral Definida [1]

Se $f$ é uma função contínua definida em $a \le x \le b$, dividimos o intervalo $[a,b]$ em $n$ subintervalos de comprimentos iguais $\Delta x = (b-a)/n$. Sejam $x_0(=a), x_1, \dots, x_n(=b)$ as extremidades desses subintervalos, e sejam $x^*_1,\dots,x^*_n$ pontos amostrais arbitrários nesses subintervalos, com $x^*_i \in [x_{i-1},x_i]$. Então a integral definida de $f$ de $a$ a $b$ é

$$
\int_a^b f(x)\,dx = \lim_{n\to\infty} \sum_{i=1}^n f(x^*_i)\,\Delta x
$$

![stewart](figs/stewart_5_11)

Fonte: https://pt.wikipedia.org/wiki/Integral

---

## Teorema Fundamental do Cálculo

Pelo Teorema Fundamental do Cálculo:

$$
\int_a^b f(x)\,dx = F(b)-F(a)
$$

onde $F$ é qualquer primitiva de $f$, isto é, $F' = f$.

Integrais comuns:

$$
\int k \,dx = kx + C, \qquad \int x^n \,dx = \frac{x^{n+1}}{n+1} + C
$$

Propriedade útil:

$$
\int_a^a f(x)\,dx = 0
$$

## Função densidade de probabilidade

Não podemos atribuir probabilidades a valores específicos, pois há uma quantidade **não enumerável** de valores. Atribuímos probabilidades a intervalos por meio de uma função; portanto, as probabilidades são representadas por áreas.

![densidade](figs/dens-1)

---

## Função densidade de probabilidade — Definição

A **função densidade de probabilidade** (fdp) atribui probabilidades a intervalos do tipo $[a,b]$ e é definida por

$$
P[a < X < b] = \int_{a}^{b} f(x)\,dx
$$

Propriedades:

i. É uma função não negativa: $f(x) \ge 0$  
ii. A área total sob a curva é igual a 1: $\displaystyle \int_{-\infty}^{+\infty} f(x)\,dx = 1$

---

## Observações sobre fdp

- $P[X=x] = 0$, portanto:  
  $P[a \le X \le b] = P[a < X \le b] = P[a \le X < b] = P[a < X < b]$  
- Qualquer função $f(\cdot)$ não negativa e com área total 1 caracteriza uma VA contínua.  
- $f(x)$ **não** representa a probabilidade de ocorrência de um ponto — a área entre dois pontos fornece a probabilidade.

## Medidas de posição para VAs contínuas

- Valor esperado (média): $\displaystyle E(X)=\mu=\int_{-\infty}^{\infty} x\,f(x)\,dx.$  
- Mediana $Md$ tem: $P(X\ge Md)\ge 1/2$ e $P(X\le Md)\ge 1/2$.  
- Moda $Mo$ é tal que $f(Mo)=\max_x f(x)$.

---

## Variância para VAs contínuas

- $\displaystyle Var(X)=\sigma^2=\int_{-\infty}^{\infty} (x-\mu)^2\,f(x)\,dx.$  
- Alternativamente: $Var(X)=E(X^2)-E(X)^2$, onde $\displaystyle E(X^2)=\int_{-\infty}^{\infty} x^2\,f(x)\,dx.$

## Principais modelos contínuos

### Modelo Uniforme contínuo

Definição: $X$ tem distribuição uniforme contínua em $[a,b]$ ($a<b$) com densidade

$$
f(x)=
\begin{cases}
\frac{1}{b-a}, & a \le x \le b\\
0, & \text{caso contrário}
\end{cases}
$$

Notação: $X\sim U[a,b]$  
Esperança: $E(X)=\dfrac{a+b}{2}$  
Variância: $Var(X)=\dfrac{(b-a)^2}{12}$

### Modelo Exponencial

Definição: $X\ge0$ segue Exponencial com parâmetro $\alpha>0$ se

$$
f(x)=
\begin{cases}
\alpha e^{-\alpha x}, & x\ge0\\
0, & \text{caso contrário}
\end{cases}
$$

Notação: $X\sim\text{Exp}(\alpha)$  
Esperança: $E(X)=1/\alpha$  
Variância: $Var(X)=1/\alpha^2$

Propriedade:

$$
P(a<X<b)=\int_a^b \alpha e^{-\alpha x}\,dx = e^{-\alpha a}-e^{-\alpha b}.
$$

![exponencial](figs/unnamed-chunk_4_1)

### Exemplo 6.6

Uma indústria fabrica lâmpadas cuja vida útil segue $\text{Exp}(1/8000)$.

- Duração média: $E(X)=8000$ horas.  
- Proporção de troca (duração < 50):

$$
P(X<50)=\int_0^{50}\frac{1}{8000}e^{-x/8000}\,dx = 1 - e^{-50/8000} \approx 0.006.
$$

### Modelo Normal

Definição: $X$ segue Normal com média $\mu$ e desvio $\sigma$ se

$$
f(x)=\frac{1}{\sigma\sqrt{2\pi}}\exp\!\left[-\frac{1}{2}\left(\frac{x-\mu}{\sigma}\right)^2\right],\quad -\infty<x<\infty.
$$

Notação: $X\sim N(\mu,\sigma^2)$  
$E(X)=\mu$, $Var(X)=\sigma^2$

![normal](figs/unnamed-chunk_5_1)

---

### Características da curva normal

- Simétrica em relação a $\mu$.  
- Moda em $x=\mu$.  
- Pontos de inflexão: $\mu\pm\sigma$.  
- Área total = 1.

Relações úteis:

$$
\begin{aligned}
&P[X>\mu]=P[X<\mu],\\
&P[\mu-\sigma<X<\mu+\sigma]\approx 0.683,\\
&P[\mu-2\sigma<X<\mu+2\sigma]\approx 0.954,\\
&P[\mu-3\sigma<X<\mu+3\sigma]\approx 0.997.
\end{aligned}
$$

---

### Probabilidades na Normal padrão

Para $Z\sim N(0,1)$:

$$
P[a<Z<b]=\int_a^b \frac{1}{\sqrt{2\pi}}e^{-z^2/2}\,dz.
$$

Transformação para qualquer $X\sim N(\mu,\sigma^2)$:

$$
P[a<X<b]=P\!\left(\frac{a-\mu}{\sigma}<Z<\frac{b-\mu}{\sigma}\right).
$$

### Normal como aproximação da Binomial

Se $X\sim\text{Bin}(n,p)$ então $E(X)=np$, $Var(X)=np(1-p)$. Aproximação por $Y\sim N(np,np(1-p))$ quando $np\ge5$ e $np(1-p)\ge5$.

### Exemplo 6.11

Com $X\sim\text{Bin}(200,0.3)$, qual $P(X\ge50)$? A probabilidade exata é difícil sem computador.

Aproximando: $Y\sim N(60,42)$ e usando correção de continuidade,

$$
P(X\ge50)\approx P\left(Y\ge49.5\right)=P\left(Z\ge\frac{49.5-60}{\sqrt{42}}\right)=P(Z\ge-1.62)\approx0.9474.
$$

### Combinação linear de Normais independentes

Se $X_i\sim N(\mu_i,\sigma_i^2)$ independentes e $a_i$ constantes, então

$$
W=\sum_{i=1}^n a_i X_i \sim N\Big(\sum_{i=1}^n a_i\mu_i,\;\sum_{i=1}^n a_i^2\sigma_i^2\Big).
$$

### Exemplo 6.13

Lucro diário (em milhares): $L=2L_A+5L_I+3L_C$ com $L_A\sim N(3,4)$, $L_I\sim N(6,9)$, $L_C\sim N(4,16)$, independentes. Qual $P(L>50)$?

$$
\begin{aligned}
\mu_L&=2\cdot3+5\cdot6+3\cdot4=48,\\
\sigma_L^2&=2^2\cdot4+5^2\cdot9+3^2\cdot16=385.
\end{aligned}
$$

Portanto $L\sim N(48,385)$ e

$$
P(L>50)=P\!\left(Z>\frac{50-48}{\sqrt{385}}\right)=P(Z>0.10)\approx0.4602.
$$
